### Wordle Question
Every day, the New York Times game [Wordle](https://en.wikipedia.org/wiki/Wordle) choses a five letter word from a list of 2,309 words that players have six attempts to guess correctly. After each attempt, the player recieves feedback on whether the letters they used are in the word and in the right location. For this question, you will be simulating the probability that the word is guessed correctly on the first attempt and that multiple player guess the same word.

The three scenarios that you will simulate are:
1. A player guesses the word on their first attempt assuming their choice is random.
2. One or more people in the same class of 45 people randomly guess the right word assumming they randomly pick six different words.
3. Two or more people in the same class of 45 people randomly guess the same word on their first attempt.

Each simulation should run 10000 times. Assign your probabilities for the three scenarios to `scen1`, `scen2`, and `scen3` respectively.

**Hint 1**: You can use an array of integers to represent the words <br>
**Hint 2**: Use replace = False in the random choice function to ensure the same word is not picked twice
